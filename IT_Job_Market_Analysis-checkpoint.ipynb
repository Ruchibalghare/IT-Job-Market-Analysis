```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("DataAnalyst.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.drop("Unnamed: 0", axis=1)
df.head()
df.columns
```

```python
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
df["Salary Estimate"].head()
```

```python
df["Salary Estimate"] = df["Salary Estimate"].str.replace("K", "")
df["Salary Estimate"] = df["Salary Estimate"].str.replace("$", "")
df["Salary Estimate"] = df["Salary Estimate"].str.replace("(Glassdoor est.)", "", regex=False)
```

```python
df[["Min Salary", "Max Salary"]] = df["Salary Estimate"].str.split("-", expand=True)
```

```python
df["Min Salary"] = pd.to_numeric(df["Min Salary"], errors='coerce')
df["Max Salary"] = pd.to_numeric(df["Max Salary"], errors='coerce')
```

```python
df["Average Salary"] = (df["Min Salary"] + df["Max Salary"]) / 2
```

```python
df[["Min Salary", "Max Salary", "Average Salary"]].head()
```

```python
df = df.drop_duplicates()
```

```python
skill_list = [
    "python", "java", "sql", "excel", "tableau", "power bi",
    "machine learning", "aws", "azure", "r", "spark",
    "hadoop", "statistics", "tensorflow", "pandas",
    "numpy", "data visualization"
]
```

```python
from collections import Counter

skill_counter = Counter()

for description in df["Job Description"].dropna():
    description = description.lower()
    for skill in skill_list:
        if skill in description:
            skill_counter[skill] += 1

skill_counter
```

```python
skill_df = pd.DataFrame(skill_counter.items(), columns=["Skill", "Count"])
skill_df = skill_df.sort_values(by="Count", ascending=False)
skill_df
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x="Count", y="Skill", data=skill_df)
plt.title("Top Technical Skills in Job Market")
plt.show()
```
